## FAISS

```python
from langchain_community.document_loaders import TextLoader 
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter


loader = TextLoader(r"C:\D\old_laptop_stuff\DL4CV\gen_ai\GenAI\RAG\speach.txt")
document = loader.load()

text_spliter = CharacterTextSplitter(chunk_size = 200, chunk_overlap = 20)
docs = text_spliter.split_documents(document)
```

```python
docs
```

```python
embedding = OllamaEmbeddings(model = "mxbai-embed-large")

db = FAISS.from_documents(docs, embedding)
db
```

### Query

```python
query = "what does the speaker belive is the main reason the united states should enter th war?"

docs = db.similarity_search(query)
docs
```

```python
docs[0].page_content
```

### Retrievers

```python
retriver = db.as_retriever()
docs = retriver.invoke(query)
docs[0].page_content
```

### similarity with score

```python
docs_and_score = db.similarity_search_with_score(query)
docs_and_score[0]
```

### save and loading

```python
db.save_local("Faiss_db")
```

```python
new_db = FAISS.load_local("Faiss_db", embedding, allow_dangerous_deserialization=True)
```

```python
new_db.similarity_search(query)
```

## ChromaDB

```python
from langchain_community.document_loaders import TextLoader 
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter


loader = TextLoader(r"C:\D\old_laptop_stuff\DL4CV\gen_ai\GenAI\RAG\speach.txt")
document = loader.load()

text_spliter = CharacterTextSplitter(chunk_size = 200, chunk_overlap = 20)
docs = text_spliter.split_documents(document)

embedding = OllamaEmbeddings(model = "mxbai-embed-large")
```

```python
## building vector db

from langchain_chroma import Chroma

vector_db = Chroma.from_documents(docs, embedding)
```
