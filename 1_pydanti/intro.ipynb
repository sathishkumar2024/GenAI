# 2. Model with Optional Fields
1. we add optinal type variables
2. the float variable automatically converts the int value to a float value

```python
from typing import Optional

class Employee(BaseModel):
    id: int
    department: str
    salary: Optional[float] = None
    isactive: Optional[bool] = True
    
emp1 = Employee(id = 1, department= "IT", salary = 100)
print(emp1)
```

```python
emp1 = Employee(id = 1, department= "IT", salary = 100, isactive=False)
print(emp1)
```

```python
class Students(BaseModel):
    room_number : str
    students : list[str]
    capacity : int
    
classroom1 = Students(room_number = "abc",students =  {"a", "b"}, capacity = 10)
print(classroom1)
```

```python
try:
    classroom1 = Students(room_number = "abc",students =  "a", capacity = 10)
    print(classroom1)

except ValueError as e:
    print(e)
```

```python
from pydantic import BaseModel, Field

class Item(BaseModel):
    name:str= Field(min_length=2, max_length=20, description="enter your name")
    price:int= Field(gt=0, le=1000, description= "it you salary")
    
item = Item(name="sss", price=10)
    

print(item)

print(item.model_json_schema())
```
